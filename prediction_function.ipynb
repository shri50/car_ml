```python
import pandas as pd

def predict_price(Brand, 
                Variant, 
                Model_Year, 
                City, 
                Kilometer, 
                Fuel_Type, 
                Drive_Mode,
                CURRENT_YEAR = 2026
                ):
    input_data = pd.DataFrame({
                 'Brand': [Brand],
                 'Variant': [Variant],
                 'Model Year': [Model_Year],
                 'City': [City],
                 'Kilometer': [Kilometer],
                 'Fuel_Type': [Fuel_Type],
                 'Drive_Mode': [Drive_Mode],
        
    })
    input_data["Car_Age"] = CURRENT_YEAR - input_data["Model Year"]
    input_data["km_per_year"] = round(input_data["Kilometer"] / input_data["Car_Age"] ,1)

    predict = model.predict(input_data)[0]
    return  predict   
```

```python
price = predict_price (
                     Brand ='Volkswagen',
                     Variant = 'Polo COMFORTLINE 1.2L PETROL',
                     Model_Year = 2023,
                     City = 'New-delhi',
                     Kilometer = 21775,
                     Fuel_Type = 'Petrol',
                     Drive_Mode = 'Manual',
                     CURRENT_YEAR = 2026
                )
price
```
